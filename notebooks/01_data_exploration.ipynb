# MahaTraffic AI — 01: Exploratory Data Analysis
**Project:** MahaTraffic AI: Road-Risk Intelligence System
**Focus:** Historical Maharashtra Accident Patterns (2019–2023) & Social Media Road-Safety Perception
*Runs locally with Pandas & PyArrow in the project virtual environment.*


```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

# Set project root
BASE_DIR = Path("..").resolve() if Path("..").resolve().name == "MahaTrafficAI" else Path(".").resolve()
print(f"Base Directory: {BASE_DIR}")
```

### 1. Load Processed Accident Parquet Dataset
We inspect the cleaned, columnar dataset stored in Parquet format.


```python
accidents_parquet = BASE_DIR / "data" / "processed" / "parquet" / "accidents" / "maharashtra_accidents_clean.parquet"
df_acc = pd.read_parquet(accidents_parquet)
print(f"Loaded {len(df_acc)} records. Columns: {list(df_acc.columns)}")
df_acc.head()
```

### 2. Yearly Accident Trends in Maharashtra
Aggregate total accidents, fatalities, injuries, and compute fatality rate per year.


```python
yearly_summary = df_acc.groupby("year").agg(
    total_accidents=("accident_count", "sum"),
    fatal_accidents=("fatal_accidents", "sum"),
    total_deaths=("deaths", "sum"),
    total_injuries=("injuries", "sum")
).reset_index()

yearly_summary["fatality_rate_%"] = (yearly_summary["total_deaths"] / yearly_summary["total_accidents"] * 100).round(2)
yearly_summary
```

### 3. Road Type Breakdown
Examine accident severity across National Highways, State Highways, and Urban Roads.


```python
road_summary = df_acc.groupby("road_type").agg(
    total_accidents=("accident_count", "sum"),
    total_deaths=("deaths", "sum"),
    total_injuries=("injuries", "sum")
).reset_index()

road_summary["fatality_rate_%"] = (road_summary["total_deaths"] / road_summary["total_accidents"] * 100).round(2)
road_summary.sort_values(by="total_accidents", ascending=False)
```

### 4. Top 10 Districts by Accident Count
Identify historical hot-spots across Maharashtra's 34 districts.


```python
district_summary = df_acc.groupby("district").agg(
    total_accidents=("accident_count", "sum"),
    total_deaths=("deaths", "sum"),
    total_injuries=("injuries", "sum")
).reset_index().sort_values(by="total_accidents", ascending=False)

district_summary.head(10)
```

### 5. Social Media Perception Dataset
Inspect public road-safety discussions, sentiment distribution, and engagement metrics.


```python
social_parquet = BASE_DIR / "data" / "processed" / "parquet" / "social_media" / "maharashtra_social_clean.parquet"
df_social = pd.read_parquet(social_parquet)
print(f"Loaded {len(df_social)} social posts.")
print("Category tag distribution:")
print(df_social["category_tag"].value_counts())
df_social.head(5)
```
